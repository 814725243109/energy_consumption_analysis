```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual aesthetic
sns.set_theme(style="whitegrid")
plt.rcParams['font.size'] = 10
```

```python
import os

# Folder create panrom
os.makedirs('/content/dataset', exist_ok=True)
file_path = '/content/dataset/Online_Retail_Cleaned_Preprocessed.xlsx'

# Check if file exists, else generate sample data automatically
if os.path.exists(file_path):
    print("Reading file from uploaded path...")
    df = pd.read_excel(file_path)
else:
    print("File not found! Generating sample dataset for analysis...")
    np.random.seed(42)
    n_rows = 500
    dates = pd.date_range(start='2023-01-01', periods=n_rows, freq='h')

    data = {
        'InvoiceNo': np.random.randint(536365, 581587, n_rows),
        'StockCode': np.random.choice(['85123A', '71053', '84406B', '84029G', '22752'], n_rows),
        'Description': np.random.choice([
            'WHITE HANGING HEART T-LIGHT HOLDER',
            'WHITE METAL LANTERN',
            'CREAM CUPID HEARTS COAT HANGER',
            'KNITTED UNION FLAG HOT WATER BOTTLE',
            'SET 7 BABUSHKA NESTING BOXES'
        ], n_rows),
        'Quantity': np.random.randint(1, 50, n_rows),
        'InvoiceDate': dates,
        'UnitPrice': np.round(np.random.uniform(1.0, 15.0, n_rows), 2),
        'CustomerID': np.random.randint(12346, 18287, n_rows),
        'Country': np.random.choice(['United Kingdom', 'Germany', 'France', 'EIRE', 'Spain', 'Netherlands'], n_rows)
    }
    df = pd.DataFrame(data)

# Calculate Total Amount column
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

# Quick Data Check
print("\n--- Data Summary ---")
print("Dataset Shape:", df.shape)
display(df.head())
```

```python
# Check Missing Values & Duplicates
print("Missing values in dataset:")
print(df.isnull().sum())

print("\nDuplicate rows count:", df.duplicated().sum())

# Ensure Date Column is Datetime format
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
# Grid layout set panrom
plt.figure(figsize=(16, 18))

# 1. Top Products by Quantity
plt.subplot(3, 2, 1)
top_products = df.groupby('Description')['Quantity'].sum().nlargest(10).reset_index()
sns.barplot(data=top_products, y='Description', x='Quantity', palette='Blues_r')
plt.title('1. Top Products by Quantity', fontsize=12, fontweight='bold')

# 2. Monthly Sales Trend
plt.subplot(3, 2, 2)
monthly_sales = df.groupby(df['InvoiceDate'].dt.to_period('M'))['TotalAmount'].sum().reset_index()
monthly_sales['InvoiceDate'] = monthly_sales['InvoiceDate'].astype(str)
sns.lineplot(data=monthly_sales, x='InvoiceDate', y='TotalAmount', marker='o', color='#1f77b4', linewidth=2.5)
plt.title('2. Monthly Sales Trend', fontsize=12, fontweight='bold')
plt.xticks(rotation=45)

# 3. Top Countries by Sales
plt.subplot(3, 2, 3)
top_countries = df.groupby('Country')['TotalAmount'].sum().nlargest(8).reset_index()
sns.barplot(data=top_countries, x='Country', y='TotalAmount', palette='Blues_r')
plt.title('3. Top Countries by Sales', fontsize=12, fontweight='bold')
plt.xticks(rotation=30)

# 4. Top Customers by Revenue
plt.subplot(3, 2, 4)
top_customers = df.groupby('CustomerID')['TotalAmount'].sum().nlargest(10).reset_index()
top_customers['CustomerID'] = top_customers['CustomerID'].astype(str)
sns.barplot(data=top_customers, x='CustomerID', y='TotalAmount', palette='Purples_r')
plt.title('4. Top Customers by Sales', fontsize=12, fontweight='bold')
plt.xticks(rotation=45)

# 5. Correlation Heatmap
plt.subplot(3, 2, 5)
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('5. Correlation Heatmap', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('visualizations_summary.png', dpi=300, bbox_inches='tight')
plt.show()
```
